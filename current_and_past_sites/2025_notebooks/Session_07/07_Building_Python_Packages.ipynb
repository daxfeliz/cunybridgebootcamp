# Setting up Project File Structure


## For a basic Project in python, the following file structure tree should look something like:
```bash
|-- ./
|-- LICENSE
|-- README.md
|-- docs/
|   |-- index.md
|-- pyproject.toml
|-- src/
|   |-- nameofproject/
|   |   |-- __init__.py
|   |   |-- cli.py
|   |   |-- core.py
|-- test/
|   |-- test_main.py
```

A useful __alias__ to copy/paste into your ~/.alias file is this one to replicate a file structure tree text output with a command __tree__:
```bash
tree() { local root="${1:-.}"; (cd "$root" && { { find . -type d -print | sed -E 's#^\./##; s#$#/#'; find . -type f -print | sed -E 's#^\./##'; } | sed '/^$/d' | sort; }) | awk -F/ '{raw=$0; isdir=(raw ~ /\/$/); sub(/\/$/,"",raw); n=split(raw,parts,"/"); for(i=1;i<n;i++) printf "|   "; printf "|-- %s%s\n", parts[n], (isdir?"/":""); }'; }
```

I recommend creating a global variable first for your project name:
```bash
project_name=nameofproject #no spaces! Name it something short, sweet, and all lowercase (to adhere to PEP8 standards)

echo ${project_name}
```

Then create the file tree with this set of commands:
```bash
mkdir docs src test src/${project_name}

touch LICENSE README.md docs/index.md pyproject.toml src/${project_name}/__init__.py src/${project_name}/cli.py src/${project_name}/core.py test/test_main.py

tree
```

and you should see an output like 
```bash
|-- ./
|-- LICENSE
|-- README.md
|-- docs/
|   |-- index.md
|-- pyproject.toml
|-- src/
|   |-- nameofproject/
|   |   |-- __init__.py
|   |   |-- cli.py
|   |   |-- core.py
|-- test/
|   |-- test_main.py
```

# Setting up pyproject.toml


## I'd recommend settings like this for this tutorial:

```bash
[project]
name = "nameofproject"
version = "0.0.1"
description = "Brief description of this tool."
authors = [ 
    { name = "FirstName LastName", email = "email@email.com" }
]
readme = "README.md"
classifiers = [ 
    "Programming Language :: Python :: 3",
    "Operating System :: OS Independent",
]
requires-python = ">=3.12"

# BELOW ARE EXAMPLE DEPENDENCIES, YOURS MIGHT NOT NEED THESE EXACT ONES
# THINK ABOUT WHAT YOU'LL NEED FOR YOUR PACKAGE
# ALSO THINK ABOUT IF THEY REQUIRE SPECIFIC VERSIONS TO WORK
dependencies = [ 
    "numpy",
    "matplotlib",
    "pandas",
    "scipy",
    "astropy",
    "astroquery"
]

[project.license]
file = "LICENSE"

[project.scripts]
# If your package folder is "get_pkmn_color_pallete", this must match:
nameofproject = "nameofproject.cli:main"

[build-system]
requires = ["setuptools", "wheel"]
build-backend="setuptools.build_meta"


[tool.setuptools]
package-dir = {"" = "src"}

[tool.setuptools.packages.find]
where = ["src"]
include = ["nameofproject*"]

```

# Common failure causes (fast mental checklist)

- Package folder missing or misnamed: must be src/nameofproject/ with __init__.py.

- include = [...] doesn’t match the folder name -> metadata-only wheel.

- Wrong import inside package (from core import * instead of from .core import *).

- Old version installed, new wheel ignored → use --force-reinstall or bump version.

- TestPyPI only used for deps → add --extra-index-url https://pypi.org/simple.

- Command name vs module path mismatch in [project.scripts]:

- Left side (command): any hyphenated name you want.

- Right side: nameofproject.cli:main (valid Python import path + callable).

- Keep this checklist; it’s the fastest end-to-end sanity pass for Python packaging with src/ layout.
