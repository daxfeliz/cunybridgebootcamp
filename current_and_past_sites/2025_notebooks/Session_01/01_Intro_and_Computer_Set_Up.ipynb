# Step 1: Installing Anaconda!

## Quick note about Licensing changes to conda:
Since 2024, Anaconda clarified its Terms of Service so that organizations with $\ge$ 200 employees or contractors must buy a paid Business license to use Anaconda’s repositories (the default “defaults” channel many conda installs use); this threshold applies to companies, governments, and non-profits, with separate carve-outs for accredited educational use (like universities) in course curricula. Practically, if your conda/Miniconda setup downloads from Anaconda’s repos inside a large organization, you’re expected to pay—or switch to a community source like conda-forge via Miniforge. More details about the changes in licensing can be found  [__here__](https://www.datacamp.com/blog/navigating-anaconda-licensing).

To combat this, institutions like Flatiron Institute Center for Computational Astrophysics (a non-profit organization) has a firewall blocking downloads from conda's default channel. You can use default conda as a package manager but you'll have to download packages from conda-forge if using conda while at CCA. Something to think about.

# Step 2: Some quick bash basics!

### Basic bash commands
* `pwd` - Returns the pathway that you are currently in (i.e. how to list where you are currently located in your machine.)
* `cd [path]` - Change the path you are in
* `ls [file name]` - list all files in the current location with a input file name
* `vim [file name]` - will open a text editor for a input file name

### vim basics:
* `vim [file name]` will activate a text editor in your terminal
* once the text editor is activated, you will see the contents of the file
* to make edits, press `"i"` once to enter "insert" mode. Then you can type in changes to the file.
* Note: cursor clicks don't work in vim, you'll need to use the arrow keys to get around. You can scroll though!
* Once your edits are complete, press `"esc"` once to exit "insert mode".
* To save your changes, type in `":w"`
* To quit the vim text editor, type in `":q"`. Note unsaved changes will prevent this.
* To force quit without saving changes, type in `":q!"`.
* To save your changes and exit the vim text editor in one step, type in `":wq"`
* To overwrite the file and force quit to exit the vim text editor in one step,, type in `":wq!"` 


# Step 3: Using Vim for editing our computer environments

After the Anaconda installation is complete, you should see that your "_PATH_" was modified in your "__~/.bashrc__" (Linux) or "__~/bash_profile__" file. If your Mac has Catalina (10.15) / Big Sur (11) / Monterey (12) / Ventura (13) / Sonoma (14) / Sequoia (15) as its operating system, it uses a "zshell" instead of a "bash shell". Instead of a "__~/.bashrc__" file, you should have a "__~/.zshrc__" file.

To check if you have a __~/.bashrc__ file (or "__~/.zshrc__" file), enter this command in your terminal:
```bash
ls ~/.bashrc
```

or 

```bash
ls ~/.zshrc
```

If you have NEITHER a __~/.bashrc__ file (or "__~/.zshrc__" file), you can create one like so:

If Mac OS $\lt$ 10.15:
```bash
vi ~/.bashrc
```

or 

If Mac OS $\ge$ 10.15:
```bash
vi ~/.zshrc
```

which will open a blank file. Enter into insert mode (type "i"), then paste in:

```bash
PATH=$PATH:~/bin
```


For Mac OSX Users, another step is required:

For users with Mac OS $\lt$ 10.15, you'll have to edit your "__~/.bash_profile__" file to include the line:

```bash
source ~/.bashrc
```

or if you have Mac OS $\ge$ 10.15, you'll have to edit your "__~/.zprofile__" file to include the line:
```bash
source ~/.zshrc
```
#### So then, in your __~/.bashrc__ file (or "__~/.zshrc__" file), you should see the following line somewhere in the file's contents:


If using miniforge:
```bash
PATH="/Users/YOUR_USER_NAME/miniforge3/bin:$PATH
```

If using anaconda:
```bash
export PATH="~/anaconda3/bin:$PATH"
```

# Step 4: Setting up Git and GitHub

Nowadays it is really important to keep track of every change that make to a code, especially if you're 
part of a large organizations.

Git is a tool for _version control_ of your code, i.e. you can keep track of
- any changes made to your code
- any issues with your code
- you can consult other team members about implementing a new feature
- you can restore a code to a previous version
- etc.

In a nutshell, __you should be using Git__.

_Git_ might already come pre-installed with _Anaconda_ (it does not if you use miniforge!). You can test whether it is installed on your machine by typing:
    
```bash
which git
```

And you shoud see the path to the _git_ executable.

If it is not installed or if you see a pathname which differs from something like `/Users/YOUR_USER_NAME/miniforge3/bin/git` OR `/Users/YOUR_USER_NAME/opt/anaconda3/bin/git` , you can install it with the command:
```bash
conda install git
```
Otherwise, you can install it from [here](https://git-scm.com/)
